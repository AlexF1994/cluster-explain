```python
from sklearn import datasets
from sklearn.cluster import KMeans
import numpy as np
from cxplain.xkm import XkmExplainer
```

```python
iris = datasets.load_iris()
X = iris.data
y = iris.target
```

```python
kmeans = KMeans(n_clusters=3, random_state=3).fit(X)
cluster_centers = kmeans.cluster_centers_
```

```python
predictions = kmeans.predict(X)
predictions
```

```python
l = np.bincount(predictions)
l
```

```python
within_explainer =  XkmExplainer(X,  kmeans.cluster_centers_, "within_scatter", "euclidean", predictions)
ratio_explainer = XkmExplainer(X,  kmeans.cluster_centers_, "scatter_ratio", "euclidean", predictions)
```

```python
explained_within_clustering = within_explainer.fit_explain()
explained_ratio_clustering = ratio_explainer.fit_explain()
```
