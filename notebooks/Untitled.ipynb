```python
iris_data = IrisDataset.load_and_clean_dataset(n_clusters=3)
```

```python
n_clusters = iris_data.n_clusters
X = iris_data.features
y = iris_data.targets
n_obs = iris_data.n_obs
n_features = iris_data.n_features

# fit Kmeans
kmeans = KMeans(n_clusters=n_clusters, random_state=3, n_init='auto').fit(X)
cluster_centers = kmeans.cluster_centers_
predictions = kmeans.predict(X)
```

```python
explainer = ExKMCExplainer(X, kmeans, k=n_clusters, max_leaves=2*n_clusters)
```

```python
expalnations = explainer.fit_explain()
```
